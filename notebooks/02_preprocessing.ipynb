```python
import sys
sys.path.append("..")  # agar folder src bisa diimpor dari folder notebooks

import pandas as pd
from src.preprocessing import preprocess_minimal, preprocess_penuh

# Baca data versi bersih (2.200 sampel)
df = pd.read_csv("../data/processed/emotweetid_clean.csv")

# Terapkan tiga versi pembersihan, masing-masing disimpan di kolom baru
df["teks_minimal"] = df["tweet"].apply(preprocess_minimal)
df["teks_minimal_slang"] = df["tweet"].apply(lambda t: preprocess_minimal(t, pakai_slang=True))
df["teks_penuh"] = df["tweet"].apply(preprocess_penuh)

# Cek: berapa teks yang kosong setelah dibersihkan?
for kolom in ["teks_minimal", "teks_minimal_slang", "teks_penuh"]:
    jumlah_kosong = (df[kolom].str.len() == 0).sum()
    print(f"Teks kosong di {kolom}: {jumlah_kosong}")

# Simpan hasilnya (folder data diabaikan Git)
df.to_csv("../data/processed/emotweetid_preprocessed.csv", index=False)
print("Tersimpan. Jumlah baris:", len(df))

# Tampilkan 5 contoh acak: sebelum dan sesudah
pd.set_option("display.max_colwidth", 120)
df[["tweet", "teks_minimal", "teks_penuh"]].sample(5, random_state=1)
```
